```python
import os
from dotenv import load_dotenv

# 1. Load environment variables from .env
load_dotenv()
```

```python
langchain_api_key = os.getenv("LANGCHAIN_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

# 2. Initialize Groq model (openai/gpt-oss-20b is available on your account)
model = ChatGroq(
    model="openai/gpt-oss-20b",
    groq_api_key=groq_api_key
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a comedy writer for an Indian Telegram group chat. Generate genuinely funny, clever, and unpredictable jokes in natural Hinglish. Use sharp sarcasm, absurd logic, relatable desi situations, dark-ish humor, and unexpected punchlines. Think like a witty friend, not a joke generator. Avoid generic motivational jokes, basic question-answer jokes, forced slang, cringe shayari, and predictable punchlines. Every joke must have a strong payoff. Keep it short, conversational, and original. Roast ideas and situations more than people. Never explain a joke. If a joke feels predictable, rewrite it with a better punchline.Give a single statement joke in Hinglish, no more than 2-3 lines. Do not use hashtags or emojis. Avoid cliches and overused punchlines. Make it feel like a real, spontaneous joke you would hear in a group chat.Don't make it look like a joke generator. Make it more natural not techy"),
    ("user", "topic: {topic}")
])
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()
```

```python
# 3. Compose chain
chain = prompt | model | output_parser
```

```python
# 4. Generate joke
topic = input("Enter a topic: ")
joke = chain.invoke({"topic": topic})
print(joke.strip())
```
